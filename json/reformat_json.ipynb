```python
import json

IN_PATH = "20260722_clusters.0.json"
OUT_PATH = "20260722_clusters.json"

with open(IN_PATH, "r", encoding="utf-8") as ifp:
  cdata = json.load(ifp)

with open(OUT_PATH, "w", encoding="utf-8") as ofp:
  json.dump(cdata, ofp, separators=(",",":"), sort_keys=True, ensure_ascii=False)
```
